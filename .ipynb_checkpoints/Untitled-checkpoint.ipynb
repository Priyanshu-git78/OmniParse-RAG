```python
from unstructured.partition.auto import partition
from unstructured.chunking.title import chunk_by_title
import time
import os 
import base64


file_name="test_documents"
file_to_process=[]
for root, dirs, files in os.walk(file_name):
    for file in files:
        full_path=os.path.join(root,file)
        file,format=full_path.split(".")
        if format in "docx,xlsx,pptx,pdf":
            file_to_process.append(full_path)
        else:
            print(f"{full_path}\n,format :{format}")

all_elements=[]
count=0
pdf_path=r"/mnt/hdd/Development Pratice/RAG Industry Level/test_documents/02_sales_and_revenue/regional_sales_performance_q3_fy26.xlsx"
for file in file_to_process[:1]:
    start= time.perf_counter()
    print(f"file: {file} in process ")
    elements = partition(
        filename=pdf_path,
        strategy="hi_res",
        infer_table_structure=True,
        extract_image_block_types=["Image"],
        extract_image_block_to_payload=True,
    )
    end = time.perf_counter()
    
    count+=1 
    print(f"count :{count},time:{end-start}" )
    all_elements.extend(elements)



# chunks = chunk_by_title(elements, max_characters=1000)

for i , chunk in enumerate(all_elements):
    verbose= True
    meta = chunk.metadata.to_dict()
    if verbose:
        print(f"chunk {i}: starts on page {meta.get('page_number')} chunk: {chunk}")
        print(meta.get("filename"))
        print(len(chunk.text))
        if chunk.metadata.image_base64:
            image=chunk.metadata.image_base64
            print(f"chunk image: {image[:10]}")






# def partition_documents(self, file_path: str = None):
#         path = file_path or self.pdf_path

#         if not os.path.isdir(path):
#             print(f"❌ Error: {path} not found")
#             return []   # return empty list, not None

#         files = []
#         for root, dirs, filenames in os.walk(path):
#             for f in filenames:
#                 files.append(os.path.join(root, f))

#         all_elements = []
#         for file in files:
#             try:
#                 print(f"📄 Partitioning document: {file}")
#                 elements = partition(
#                     filename=file,
#                     strategy="hi_res",
#                     infer_table_structure=True,
#                     extract_image_block_types=["Image"],
#                     extract_image_block_to_payload=True,
#                 )
#                 all_elements.extend(elements)
#             except Exception as e:
#                 print(f"⚠️ Failed to partition {file}: {e}")

#         return all_elements
```
